# Task 8
## Promote Designation Level for Employees Whose Age is More Than 29 Years

```python
# Display the Final DataFrame
final_df
```

```python
# Display employees whose Age is greater than 29
final_df[final_df["Age"] > 29]
```

```python
# Promote Designation Level by 1 for employees whose Age is greater than 29
for i in range(len(final_df)):
    if final_df.loc[i, "Age"] > 29:
        final_df.loc[i, "Designation Level"] = final_df.loc[i, "Designation Level"] - 1
```

# Task 9
## Calculate Total Project Cost for Each Employee

```python
# Display the Final DataFrame
final_df
```

```python
# Group the data by ID and First Name
final_df.groupby(["ID", "First Name"])
```
